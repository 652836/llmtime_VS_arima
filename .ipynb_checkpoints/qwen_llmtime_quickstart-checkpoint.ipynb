# Qwen LLMTime Quickstart

This notebook runs the minimum sampling-only path:

1. Load a local time series.
2. Configure an OpenAI-compatible provider.
3. Call `get_llmtime_predictions_data(..., compute_nll=False)`.
4. Deserialize the returned text samples into numeric forecasts.
5. Plot the median forecast against the ground truth.


```python
import os

import matplotlib.pyplot as plt
import pandas as pd

from data.serialize import SerializerSettings
from models.llmtime import get_llmtime_predictions_data
from models.provider_adapter import configure_model, get_model_config
```

```python
# Fill these values before you run the prediction cell.
# You can either edit this cell directly or export matching env vars first.

QWEN_LOGICAL_MODEL = "qwen-plus"

config = configure_model(
    logical_model_name=QWEN_LOGICAL_MODEL,
    provider="openai_compatible",
    api_key=os.getenv("OPENAI_COMPAT_API_KEY", "YOUR_API_KEY_HERE"),
    base_url=os.getenv("OPENAI_COMPAT_BASE_URL", "YOUR_OPENAI_COMPATIBLE_BASE_URL_HERE"),
    api_model_name="qwen-plus",
    completion_mode="chat",
    tokenizer_alias="cl100k_base",
    context_length=8192,
    supports_logprobs=False,
)

resolved_config = get_model_config(QWEN_LOGICAL_MODEL)
if (
    not resolved_config.api_key
    or resolved_config.api_key.startswith("YOUR_")
    or not resolved_config.base_url
    or resolved_config.base_url.startswith("YOUR_")
):
    raise ValueError(
        "Fill api_key and base_url in this cell before running the prediction cell."
    )

resolved_config
```

```python
df = pd.read_csv("AirPassengers.csv")
series = pd.Series(df["Passengers"].astype(float).values, index=pd.RangeIndex(len(df)))

forecast_horizon = 24
train = series.iloc[:-forecast_horizon]
test = series.iloc[-forecast_horizon:]

settings = SerializerSettings(
    base=10,
    prec=3,
    signed=True,
    half_bin_correction=True,
)

print(f"train length: {len(train)}")
print(f"test length: {len(test)}")

plt.figure(figsize=(10, 4))
plt.plot(train.index, train.values, label="train")
plt.plot(test.index, test.values, label="test", color="black")
plt.legend()
plt.title("AirPassengers split")
plt.show()
```

```python
preds = get_llmtime_predictions_data(
    train,
    test,
    model=QWEN_LOGICAL_MODEL,
    settings=settings,
    num_samples=5,
    temp=0.7,
    alpha=0.95,
    beta=0.3,
    basic=False,
    parallel=False,
    compute_nll=False,
)

preds.keys()
```

```python
median = pd.Series(preds["median"], index=test.index)

plt.figure(figsize=(10, 4))
plt.plot(train.index, train.values, label="train")
plt.plot(test.index, test.values, label="truth", color="black")
plt.plot(median.index, median.values, label="Qwen median", color="tab:orange")
plt.legend()
plt.title("LLMTime sampling forecast")
plt.show()

if preds["completions_list"] is not None:
    print("First raw completion sample:")
    print(preds["completions_list"][0][0])

preds["samples"].head()
```
